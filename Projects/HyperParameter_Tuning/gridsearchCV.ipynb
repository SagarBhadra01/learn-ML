```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
df.head()
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split
X = df.drop('species', axis=1)
y = df['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model_knn = KNeighborsClassifier(n_neighbors=10)
```

```python
model_knn.fit(X_train, y_train)
```

```python
model_knn.score(X_test, y_test)
```

```python
from sklearn.svm import SVC
```

```python
model_svm = SVC(gamma='auto')
```

```python
model_svm.fit(X_train, y_train)
```

```python
model_svm.score(X_test, y_test)
```

Grid search CV

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifiers = GridSearchCV(estimator=model_svm, param_grid={'C': [0.1, 1, 10, 20, 30], 'kernel': ['rbf', 'linear'], 'gamma': ['scale', 'auto']}, cv=5, return_train_score=False)
```

```python
classifiers.fit(X,y)
```

```python
classifiers.cv_results_
```

```python
results = pd.DataFrame(classifiers.cv_results_)
results
```

```python
results[['param_C', 'param_kernel', 'param_gamma', 'mean_test_score']]
```
